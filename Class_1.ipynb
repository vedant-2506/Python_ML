```python
#3
num = []

n = int(input("Enter the size of list "))

even = []
odd = []

for i in range (n):
    number = int(input("Enter a number: "))
    num.append(number)

print(num)

for i in range (n):
    if num[i] % 2 == 1 :
        odd.append(num[i])
    else:
        even.append(num[i])

print("Even") 
print(even)
print("Odd")
print(odd)    
```

```python
#4
num = []

n = int(input("Enter the size of list "))

for i in range (n):
    number = int(input("Enter a number: "))
    num.append(number)

print(num)

#Largest element
max = num[0]
for i in range(n):
    if num[i] > max:
        max=num[i]

print("Largest element" , max)

#Smallest element
min = num[i]
for i in range(n):
    if num[i] < min:
        min = num[i]

print("Smallest element" , min) 

#  • Sum of all elements 
sum=0
for i in range(n):
    sum+=num[i]

print("Largest element" , sum) 

# • Average
avg = sum/n
print("Average" , avg)


```
