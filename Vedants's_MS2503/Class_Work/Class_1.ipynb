```python
#4
num = []

n = int(input("Enter the size of list "))

for i in range (n):
    number = int(input("Enter a number: "))
    num.append(number)

print(num)

#Largest element
max = num[0]
for i in range(n):
    if num[i] > max:
        max=num[i]

print("Largest element" , max)

#Smallest element
min = num[i]
for i in range(n):
    if num[i] < min:
        min = num[i]

print("Smallest element" , min) 

#  • Sum of all elements 
sum=0
for i in range(n):
    sum+=num[i]

print("Sum of all elements" , sum) 

# • Average
avg = sum/n
print("Average" , avg)


```

```python
#6 
n = int(input("Enter the size of list "))
marks = ()
for i in range(n):
    mark = tuple(map(int, input("Enter marks separated by space: ").split()))
    marks += mark

#Display
print("Tuple:", marks)

#Highest mark
max=marks[0]
for i in range(n):
    if marks[i] > max:
        max=marks[i]    
print("Highest mark:", max)

#Lowest mark
min=marks[0]
for i in range(n):
    if marks[i] < min:
        min=marks[i]
print("Lowest mark:", min)

#Number of students scoring above the class average
total = 0
for i in range(n):
    total += marks[i]   
print("Total marks:", total)

average = total / n
print("Class average:", average)

count = 0
for i in range(n):
    if marks[i] > average:
        count += 1
print("Number of students scoring above the class average:", count)
```

```python
#8

string = input("Enter a string: ")

freq ={}

for ch in string:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1

print("Frequency of each character in '", string, "':")
print(freq)
```

```python
#9

set1=set()
set2=set()

#set1
n = int(input("Enter the number of elements in set 1: "))
for i in range(n):
    element = input("Enter element for set 1: ")
    set1.add(element)

#set2
m = int(input("Enter the number of elements in set 2: "))
for i in range(m):
    element = input("Enter element for set 2: ")
    set2.add(element)

#display sets
print("Set 1:", set1)
print("Set 2:", set2)

#union
union_set = set1.union(set2)
print("Union of sets:", union_set)

#Intersection
intersection_set = set1.intersection(set2)
print("Intersection of sets:", intersection_set)

# Difference
difference_set = set1.difference(set2)
print("Difference of sets (Set 1 - Set 2):", difference_set)

# Symmetric Difference
symmetric_diff_set = set1.symmetric_difference(set2)
print("Symmetric Difference of sets:", symmetric_diff_set)
```

```python
# 10. Contact Book

contact = {}

while True:

    print("\n===== CONTACT BOOK =====")
    print("1. Add Contact")
    print("2. Search Contact")
    print("3. Update Contact")
    print("4. Delete Contact")
    print("5. Display All Contacts")
    print("6. Exit")

    choice = input("Enter your choice (1-6): ")

    # Add a contact
    if choice == '1':
        name = input("Enter contact name: ")
        phone = input("Enter contact phone number: ")

        contact[name] = phone

        print("Contact added successfully.")

    # Search for a contact
    elif choice == '2':
        name = input("Enter contact name to search: ")

        if name in contact:
            print("Phone number of", name, "is", contact[name])
        else:
            print(name, "not found in contacts.")

    # Update a contact
    elif choice == '3':
        name = input("Enter contact name to update: ")

        if name in contact:
            new_phone = input("Enter new phone number: ")
            contact[name] = new_phone

            print("Contact updated successfully.")
        else:
            print(name, "not found in contacts.")

    # Delete a contact
    elif choice == '4':
        name = input("Enter contact name to delete: ")

        if name in contact:
            del contact[name]

            print("Contact deleted successfully.")
        else:
            print(name, "not found in contacts.")

    # Display all contacts
    elif choice == '5':

        if contact:
            print("\nALL CONTACTS")

            for name, phone in contact.items():
                print("Name:", name, "| Phone:", phone)

        else:
            print("No contacts available.")

    # Exit
    elif choice == '6':
        print("Thank you!")
        break

    # Invalid choice
    else:
        print("Invalid choice.")
```

```python
# 15

students = {}

while True:

    print("\n===== STUDENT RECORD MANAGEMENT SYSTEM =====")
    print("1. Add Student")
    print("2. Search Student")
    print("3. Update Student")
    print("4. Delete Student")
    print("5. Display All Students")
    print("6. Exit")

    choice = input("Enter your choice (1-6): ")

    # Add a student
    if choice == '1':

        roll_no = input("Enter Roll Number: ")

        if roll_no in students:
            print("Student already exists.")

        else:
            name = input("Enter Name: ")
            course = input("Enter Course: ")
            marks = float(input("Enter Marks: "))

            students[roll_no] = {
                "Name": name,
                "Course": course,
                "Marks": marks
            }

            print("Student added successfully.")

    # Search for a student
    elif choice == '2':

        roll_no = input("Enter Roll Number to search: ")

        if roll_no in students:
            print("\nStudent Found")
            print("Name:", students[roll_no]["Name"])
            print("Roll Number:", roll_no)
            print("Course:", students[roll_no]["Course"])
            print("Marks:", students[roll_no]["Marks"])

        else:
            print("Student not found.")

    # Update student information
    elif choice == '3':

        roll_no = input("Enter Roll Number to update: ")

        if roll_no in students:

            print("\n1. Update Name")
            print("2. Update Course")
            print("3. Update Marks")

            update_choice = input("Enter your choice: ")

            if update_choice == '1':
                students[roll_no]["Name"] = input("Enter new name: ")
                print("Name updated successfully.")

            elif update_choice == '2':
                students[roll_no]["Course"] = input("Enter new course: ")
                print("Course updated successfully.")

            elif update_choice == '3':
                students[roll_no]["Marks"] = float(input("Enter new marks: "))
                print("Marks updated successfully.")

            else:
                print("Invalid choice.")

        else:
            print("Student not found.")

    # Delete a student
    elif choice == '4':

        roll_no = input("Enter Roll Number to delete: ")

        if roll_no in students:
            del students[roll_no]
            print("Student deleted successfully.")

        else:
            print("Student not found.")

    # Display all students
    elif choice == '5':

        if len(students) == 0:
            print("No student records available.")

        else:
            print("\n===== ALL STUDENTS =====")

            for roll_no, student in students.items():

                print("\nRoll Number:", roll_no)
                print("Name:", student["Name"])
                print("Course:", student["Course"])
                print("Marks:", student["Marks"])

    # Exit
    elif choice == '6':

        print("Thank you!")
        break

    # Invalid choice
    else:
        print("Invalid choice. Please try again.")
```
