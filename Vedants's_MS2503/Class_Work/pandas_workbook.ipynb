## Understanding a DataFrame

Before we start manipulating data, let's understand the basic structure of a DataFrame.

A DataFrame consists of:

- **Rows** — individual observations or records
- **Columns** — variables or attributes
- **Index** — labels used to identify rows

For our DataFrame:

| Index | Name | Age | City | Score |

## Selecting Rows

So far, we have learned how to select columns.

But what if we want to select specific **rows**?

Pandas provides two important tools for selecting rows:

- `iloc` — select using the **position** of the row
- `loc` — select using the **label** of the row

Let's start with `iloc`.

### `iloc` — Integer Position Based Selection

`iloc` stands for **integer-location**.

It allows us to select rows and columns based on their **numerical position**.

Remember that our DataFrame has the following row positions:

| Position | Index | Name |
|---:|---:|---|
| 0 | 0 | Alice |
| 1 | 1 | Bob |
| 2 | 2 | Charlie |

To select the first row:



df.iloc[0]

````python
#code here

df.iloc[1]
````

The result is a **Series** containing the values from that row.

Just like selecting a single column:

```python
df["Name"]

#### Selecting Multiple Rows

df.iloc[0:2]

````python
#code here

df.iloc[0:2]
````

### Selecting a Specific Row and Column

`iloc` can also be used to select both rows and columns.

The general syntax is:

```python
df.iloc[row, column]

df.iloc[0, 1]

````python
#code here

df.iloc[row, column]
````

We can also select a range of rows and columns:

```python
df.iloc[0:2, 0:2]

````python
#code here 
df.iloc[0:2, 0:2]

#slect second column
df.iloc[:, 1]

#selcet last row
df.iloc[-1]

#selct 1st 2 rows and 1st 2 rows
df.iloc[0:2, 0:2]


#slect name age of 1st 2 rows
df.iloc[0:2, [0, 1]]
````

## `loc` — Label-Based Selection

`loc` is used to select data based on **labels**.

The important distinction is:

> `iloc` → **position**  
> `loc` → **label**

So far, our DataFrame has the default index:

| Index | Name | Age | City | Score |
|------:|------|----:|------|------:|
| 0 | Alice | 21 | Pune | 85 |
| 1 | Bob | 22 | Mumbai | 91 |
| 2 | Charlie | 20 | Pune | 78 |

Here, the index values (`0`, `1`, `2`) are also labels.

For example:



```python
df.loc[0]

````python
#code here
df.loc[0]
````

### `loc` vs `iloc`

With our current DataFrame, these may appear to do the same thing:

```python
df.loc[0]
