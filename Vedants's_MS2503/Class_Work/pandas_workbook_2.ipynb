### Missing Data in Real Datasets

In real-world data, it's common to encounter missing or incomplete values.

Missing data can occur due to:
- Data collection errors
- Incomplete surveys
- Equipment failures
- Data entry mistakes
- Intentional omissions

Pandas represents missing data as:
- `NaN` (Not a Number) for numerical data
- `None` for object/string data

Missing data can cause problems in analysis and calculations, so we need to handle it properly.

````python
# Let's create a DataFrame with missing data to explore
import pandas as pd
import numpy as np

# Create sample data with missing values
students_with_missing = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Python': [85, np.nan, 78, 90, np.nan],
    'Maths': [99, 85, np.nan, 50, 88],
    'Chemistry': [90, np.nan, 75, np.nan, 92],
    'Physics': [85, 65, 87, 33, np.nan]
})

print("DataFrame with Missing Values:")
print(students_with_missing)
print("\nData types:")
print(students_with_missing.dtypes)
````

### Handling Missing Data: dropna()

`dropna()` removes rows or columns containing missing values.

**Strategies:**
- Drop entire rows with ANY missing value
- Drop entire columns with ANY missing value
- Drop rows where ALL values are missing
- Drop columns where ALL values are missing

**Syntax:**
```python
df.dropna()                           # Remove rows with any NaN
df.dropna(axis=1)                     # Remove columns with any NaN
df.dropna(how='all')                  # Remove rows where ALL values are NaN
df.dropna(subset=['column_name'])     # Remove rows where specific column has NaN
df.dropna(thresh=3)                   # Keep rows with at least 3 non-null values
```

**Warning:** Dropping rows/columns removes data! Use with caution.

````python
# code here
# Try different dropna() strategies:
# 1. Remove rows with ANY missing value
students_with_missing.dropna()

# 2. Remove columns with ANY missing value
students_with_missing.dropna(axis=1)

# 3. Remove rows where ALL values are missing
students_with_missing.dropna(how="all")

# 4. Remove rows where specific columns have NaN
students_with_missing.dropna(subset=["Python", "Maths"])
````

### Handling Missing Data: fillna()

`fillna()` replaces missing values with specified values instead of removing data.

This is often better than dropping data because it preserves row/column count.

**Methods:**
```python
df.fillna(0)                          # Replace all NaN with 0
df.fillna('Unknown')                  # Replace with a specific value
df.fillna(method='ffill')             # Forward fill: use previous value
df.fillna(method='bfill')             # Backward fill: use next value
df['column'].fillna(df['column'].mean())  # Fill with mean/median/mode
df.fillna({'col1': 0, 'col2': 'N/A'}) # Fill different columns differently
```

**Forward Fill (ffill):** Propagates the last known value forward
**Backward Fill (bfill):** Propagates the next known value backward

````python
# code here
# Try different fillna() strategies:

# 1. Fill missing values with a constant value
students_with_missing.fillna(0)

# 2. Fill missing values with the mean of each column
students_with_missing.fillna(
    students_with_missing.select_dtypes(include='number').mean()
)
# 3. Use forward fill (ffill)
students_with_missing.ffill()

# 4. Fill different columns with different values
students_with_missing.fillna({
    'Python': 0,
    'Maths': 50,
    'Chemistry': 0,
    'Physics': 40
})

# 5. Use backward fill (bfill)
students_with_missing.bfill()
````

### Grouping Data with groupby()

`groupby()` splits data into groups based on one or more columns, then applies operations to each group.

This is powerful for answering questions like:
- "What is the average score per department?"
- "How many students are in each department?"
- "What is the total sales per region?"

**Basic Syntax:**
```python
df.groupby('column').operation()
```

**Common Operations:**
```python
df.groupby('Department').mean()       # Average for each group
df.groupby('Department').sum()        # Sum for each group
df.groupby('Department').count()      # Count of rows in each group
df.groupby('Department').min()        # Minimum value in each group
df.groupby('Department').max()        # Maximum value in each group
df.groupby('Department').agg(['mean', 'sum', 'count'])  # Multiple operations
```

**Grouping by Multiple Columns:**
```python
df.groupby(['Department', 'Gender']).mean()  # Nested grouping
```

````python
# Let's create a fresh DataFrame to practice groupby()
# First, let's use our original students DataFrame with the Department column

# Quick refresh - create the students DataFrame
df_students = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Python': [85, 92, 78, 90],
    'Maths': [99, 85, 88, 50],
    'Chemistry': [90, 65, 75, 80],
    'Physics': [85, 65, 87, 33],
    'Department': ['SCMS', 'SCMS', 'CS', 'Physics']
})

print(df_students)
````

````python
# code here
# Try these groupby() operations:
# 1. Group by Department and calculate mean marks
print("1. Mean marks by Department:")
print(df_students.groupby('Department')[['Python', 'Maths', 'Chemistry', 'Physics']].mean())


# 2. Group by Department and calculate sum of marks
print("\n2. Sum of marks by Department:")
print(df_students.groupby('Department')[['Python', 'Maths', 'Chemistry', 'Physics']].sum())


# 3. Group by Department and count students
print("\n3. Number of students in each Department:")
print(df_students.groupby('Department').size())


# 4. Find the department with the highest average Python score
python_mean = df_students.groupby('Department')['Python'].mean()
highest_python_department = python_mean.idxmax()

print("\n4. Department with the highest average Python score:")
print(highest_python_department)
print("Average Python score:", python_mean.max())


# 5. Use agg() to get multiple statistics at once
print("\n5. Multiple statistics by Department:")
print(df_students.groupby('Department')[['Python', 'Maths', 'Chemistry', 'Physics']].agg(['mean', 'sum', 'count']))
````
