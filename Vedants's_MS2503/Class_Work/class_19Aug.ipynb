```python
arr1 = np.array([[1, 2, 3], [4, 5, 6]])
arr2 = np.array([[7, 8, 9], [10, 11, 12]])
print("array 1:", arr1)
print("array 2:", arr2)
con = np.concatenate([arr1, arr2])
print("concatenated array:",  con)

has = np.hstack(([arr1, arr2]))
print("horizontally stacked array:", has)

print("dimensions of concatenated array:", con.ndim)
print("dimensions of horizontally stacked array:", has.ndim)
```
