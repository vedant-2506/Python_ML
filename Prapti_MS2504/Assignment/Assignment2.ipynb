Data Structures

1.List

append(),insert(),remove(),pop(),sort()

```python
fruits = ["Apple", "Banana"]

fruits.append("Mango")

print(fruits)
```

append=()= Append the new value at the end of the list.

```python
fruits = ["Apple", "Banana"]

fruits.insert(1, "Orange")

print(fruits)
```

When we want the new item to appear at a particular location in the list.

```python
fruits = ["Apple", "Banana", "Mango"]

fruits.remove("Banana")

print(fruits)
```

Remove the first occurence of a specific element from the list

```python
fruits = ["Apple", "Banana", "Mango"]

removed = fruits.pop()

print(fruits)
print(removed)
```

removes and returns an element from the list.

```python
fruits = ["Banana", "Mango", "Apple","Berry"]
fruits.sort()

print(fruits)
```

arranges the elements of the list in ascending order by default.

Tuple

count(),index(),len(),max(),min()

```python
numbers = (10, 20, 30, 20, 40)

ans=numbers.count(20)

print(ans)
```

Returns the number of times a specified element appears in the tuple.

```python
numbers = (10, 20, 30, 20, 40)

ans=numbers.index(20)

print(ans)
```

Returns the index of the first occurrence of a specified element.

```python
numbers = (10, 20, 30, 40)

ans=len(numbers)

print(ans)
```

Returns the total number of elements in the tuple.

```python
numbers = (10, 20, 30, 40)

ans=max(numbers)

print(ans)
```

Returns the largest element in the tuple.

```python
numbers = (10, 20, 30, 40)

ans=min(numbers)

print(ans)
```

Returns the smallest element in the tuple.

Set

add(),discard(),update(),union(),intersection()

```python
fruits = {"Apple", "Banana"}

fruits.add("Mango")

print(fruits)
```

To insert a new unique element into the set.

```python
fruits = {"Apple", "Banana", "Mango"}

fruits.discard("Orange")

print(fruits)
```

To remove an element without causing an error if it isn't present.

```python
fruits = {"Apple", "Banana"}
fruits1 = {"Grapes", "Banana", "Mango"}
fruits.update(fruits1)
print(fruits)
```

Adds all elements from another set to the current set.

```python
fruits = {"Apple", "Banana","Berry"}
fruits1 = {"Grapes", "Banana", "Mango"}
ans=fruits.union(fruits1)
print(ans)
```

Returns a new set containing all unique elements from both sets.

```python
fruits = {"Apple", "Banana","Berry"}
fruits1 = {"Grapes", "Banana", "Mango"}
ans=fruits.intersection(fruits1)
print(ans)
```

Returns a new set containing only the common elements.

Dictionary

get(),keys(),values(),items(),update()

```python
Fruits={"name":"Mango","Quantity":5}
ans=Fruits.get("name")
print(ans)
```

Returns the value of the specified key.

```python
Fruits={"name":"Mango","Quantity":5}
ans=Fruits.keys()
print(ans)
```

Returns all the keys in the dictionary.

```python
Fruits={"name":"Mango","Quantity":5}
ans=Fruits.values()
print(ans)
```

Returns all the values in the dictionary.

```python
Fruits={"name":"Mango","Quantity":5}
ans=Fruits.items()
print(ans)
```

Returns all key-value pairs as tuples.

```python
Fruits={"name":"Mango","Quantity":5}
Fruits.update({"Colour": "Yellow","Quantity":10})
print(Fruits)
```

Returns all key-value pairs as tuples.
