Problem 11

```python
class Student:
    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks

    def average_marks(self):
        return sum(self.marks) / len(self.marks)
student1=Student("Rahul",1,[80,75,90])
student2=Student("Priya",2,[85,90,95])
student3=Student("Amit",3,[70,80,75])

print("Rahul average Marks:",student1.average_marks())
print("Priya average Marks:",student2.average_marks())
print("Amit average Marks:",student3.average_marks())
```
