```python
m1=float(input("Enter marks of sunject 1"))
m2=float(input("Enter marks of sunject 2"))
m3=float(input("Enter marks of sunject 3"))
m4=float(input("Enter marks of sunject 4"))
m5=float(input("Enter marks of sunject 5"))
print(m1)
print(m2)
print(m3)
print(m4)
print(m5)

marks=[m1,m2,m3,m4,m5]
print(marks)

average=m1+m2+m3+m4+m5/len(marks)
print("Average:", average)
if average >= 90:
    grade = "A"
elif average >= 75:
    grade = "B"
elif average >= 60:
    grade = "C"
elif average >= 40:
    grade = "D"
else:
    grade = "F"
print("Grade:", grade)
```

Problem 4

```python
n1 = float(input("Enter number 1: "))
n2 = float(input("Enter number 2: "))
n3 = float(input("Enter number 3: "))
n4 = float(input("Enter number 4: "))
n5 = float(input("Enter number 5: "))

numbers = [n1, n2, n3, n4, n5]

largest = numbers[0]
smallest = numbers[0]

for n in numbers:
    if n > largest:
        largest = n
    if n < smallest:
        smallest = n

print("Numbers:", numbers)
print("Largest number is:", largest)
print("Smallest number is:", smallest)
print("Sum of numbers is:", n1 + n2 + n3 + n4 + n5)
print("Average is:", (n1 + n2 + n3 + n4 + n5) / len(numbers))
```

Problem 15

```python
students = {}

for i in range(10):

    print("\nStudent Record")
    print("1. Add Student")
    print("2. Search Student")
    print("3. Update Student")
    print("4. Delete Student")
    print("5. Display All Students")
    print("6. Exit")
    choice = input("Enter your choice: ")
    if choice == "1":
        roll = input("Enter Roll Number: ")
        name = input("Enter Name: ")
        course = input("Enter Course: ")
        marks = input("Enter Marks: ")
        students[roll] = [name, course, marks]
        print("Student added")

    elif choice == "2":
        roll = input("Enter Roll Number: ")
        if roll in students:
            print("Name:", students[roll][0])
            print("Course:", students[roll][1])
            print("Marks:", students[roll][2])
        else:
            print("Student not found")

    elif choice == "3":
        roll = input("Enter Roll Number: ")
        if roll in students:
            students[roll][0] = input("Enter new Name: ")
            students[roll][1] = input("Enter new Course: ")
            students[roll][2] = input("Enter new Marks: ")
            print("Student updated")
        else:
            print("Student not found")
    elif choice == "4":
        roll = input("Enter Roll Number:")

        if roll in students:
            del students[roll]
            print("Student deleted")
        else:
            print("Student not found")

    elif choice == "5":
        print("\nAll Students:")
        for roll in students:
            print("Roll Number:", roll)
            print("Name:", students[roll][0])
            print("Course:", students[roll][1])
            print("Marks:", students[roll][2])
    elif choice == "6":
        print("exit")
        break
    else:
        print("Invalid choice")
```
