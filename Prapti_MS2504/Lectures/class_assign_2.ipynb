Problem 2

```python
import numpy as np
def temperature_analysis(temperatures):
    average =np.mean(temperatures)
    maximum =np.max(temperatures)
    minimum =np.min(temperatures)
    for i in range(len(temperatures)):
        if temperatures[i] ==maximum:
            highest_day =i + 1
            break
    above_average =temperatures[temperatures > average]

    print("Average Temperature:",average)
    print("Maximum Temperature:",maximum)
    print("Minimum Temperature:", minimum)
    print("Day with Highest Temperature: Day",highest_day)
    print("Temperatures Above Average:", above_average)


temperatures = np.array([32.5,31.8,30.2,33.1,34.5,29.8,31.0])
temperature_analysis(temperatures)
```

```python
import numpy as np
marks=np.array([
    [78,82,91],
    [65,74,80],
    [92,88,95],
    [56,69,72],
    [84,90,87]
])
students=["Student 1","Student 2","Student 3","Student 4","Student 5"]
subjects = ["Mathematics","Physics","Programming"]
student_average=np.mean(marks,axis=1)

print("Average marks of each student:")
for i in range(5):
    print(students[i], ":",student_average[i])
subject_average = np.mean(marks, axis=0)
print("\nAverage marks in each subject:")
for i in range(3):
    print(subjects[i], ":",subject_average[i])

highest =np.max(marks)
lowest =np.min(marks)

print("\nHighest mark:",highest)
print("Lowest mark:",lowest)
highest_student =0

for i in range(1, 5):
    if student_average[i]>student_average[highest_student]:
        highest_student=i

print("Student with highest average:",students[highest_student])

highest_subject=0

for i in range(1, 3):
    if subject_average[i]>subject_average[highest_subject]:
        highest_subject =i

print("Subject with highest average:",subjects[highest_subject])
```

Problem 12

```python
import numpy as np
image = np.array([
    [0,0,0,255,255],
    [0,0,128,255,255],
    [0,128,128,128,255],
    [0,0,128,0,0],
    [255,0,0,0,255]
])
minimum =np.min(image)
maximum =np.max(image)
mean =np.mean(image)
greater_100=[]
equal_255=[]
for row in image:
    for pixel in row:
        if pixel > 100:
            greater_100.append(pixel)
        if pixel==255:
            equal_255.append(pixel)
greater_100=np.array(greater_100)
equal_255=np.array(equal_255)
print("Minimum pixel intensity:",minimum)
print("Maximum pixel intensity:",maximum)
print("Mean pixel intensity:",mean)
print("Number of pixels greater than 100:",len(greater_100))
print("Pixels equal to 255:")
print(equal_255)
print("New array with pixels greater than 100:")
print(greater_100)
```

Problem 14

```python
import numpy as np
def create_dataset():
    data =np.random.randint(1,101,(10, 5))
    print("Dataset:")
    print(data)
    print("\nShape:",data.shape)
    print("Dimensions:",data.ndim)
    return data
def analyze_data(data):
    variable_mean =np.mean(data,axis=0)
    print("\nMean of each variable:")
    print(variable_mean)
    experiment_mean =np.mean(data,axis=1)
    print("\nMean of each experiment:")
    print(experiment_mean)
    print("\nMaximum value:",np.max(data))
    print("Minimum value:",np.min(data))
    overall_mean =np.mean(data)
    print("Overall mean:",overall_mean)
    above_mean = []
    for row in data:
        for value in row:
            if value > overall_mean:
                above_mean.append(value)
    print("\nValues above overall mean:")
    print(np.array(above_mean))
    new_data =data.reshape(5, 10)
    print("\nReshaped data:")
    print(new_data)
    print("New shape:",new_data.shape)
    flattened_data =new_data.reshape(-1)
    print("\nFlattened data:")
    print(flattened_data)
    print("Flattened shape:",flattened_data.shape)
data =create_dataset()

analyze_data(data)
```

Problem 15

```python
import numpy as np
def generate_marks():
    marks =np.random.randint(0,101,(10, 4))
    return marks
def student_averages(marks):
    return np.mean(marks,axis=1)
def subject_averages(marks):
    return np.mean(marks,axis=0)
def get_grade(average):
    if average>=90:
        return "A+"
    elif average>=80:
        return "A"
    elif average>=70:
        return "B"
    elif average>= 60:
        return "C"
    elif average>=50:
        return "D"
    else:
        return "F"
def create_student_information(averages):
    students =[]
    for i in range(10):
        grade =get_grade(averages[i])
        student = {
            "Student": i + 1,
            "Average": averages[i],
            "Grade": grade
        }
        students.append(student)
    return students
marks =generate_marks()
print("========== STUDENT PERFORMANCE ANALYSIS ==========\n")
print("Marks of Students:")
print(marks)
student_avg =student_averages(marks)
print("\nAverage Mark of Each Student:")
for i in range(10):
    print("Student", i + 1, ":", round(student_avg[i], 2))

subject_avg =subject_averages(marks)
print("\nAverage Mark of Each Subject:")
for i in range(4):
    print("Subject", i + 1, ":",round(subject_avg[i], 2))

highest_student=0
lowest_student=0
for i in range(1,10):
    if student_avg[i]>student_avg[highest_student]:
        highest_student = i

    if student_avg[i]<student_avg[lowest_student]:
        lowest_student=i
print("\nStudent with Highest Average:")
print("Student",highest_student + 1,":",round(student_avg[highest_student],2))
print("\nStudent with Lowest Average:")
print("Student",lowest_student + 1,":",round(student_avg[lowest_student],2))
highest_mark =np.max(marks)
lowest_mark =np.min(marks)
print("\nHighest Mark in Entire Dataset:",highest_mark)
print("Lowest Mark in Entire Dataset:",lowest_mark)
count = 0
for average in student_avg:
    if average >75:
        count+=1
print("\nStudents with Average Above 75:",count)
marks_above_90 =marks[marks > 90]
print("\nAll Marks Greater Than 90:")
print(marks_above_90)
student_information=create_student_information(student_avg)
print("\n========== FINAL STUDENT INFORMATION ==========")
for student in student_information:
    print("Student:", student["Student"],"| Average:", round(student["Average"], 2),"| Grade:", student["Grade"] )
```
